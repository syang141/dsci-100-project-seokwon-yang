#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:

- Transform the data into a tidy form (if needed)

```r
march_tidy <- march_madness |>
# we only keep 1994 and afterwards: a lot more teams starting 1994 &
# seed(1-16) and tournament wins (max 6) are on the same scale every year 
    filter(year >= 1994) |>
# remove rows with NA in the variables we want
    filter(!is.na(seed), !is.na(reg_wins_pct), !is.na(tourney_wins), !is.na(bid)) |>
# keep identifiers (year, school) and predictors (seed, reg_wins_pct)
# the response (tourney_wins) and the required categorical value (bid)
    select(year, school, seed, bid, reg_wins_pct, tourney_wins) |>
# convert bid to a factor
    mutate(bid = as_factor(bid))

head(march_tidy, 6)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
# First step is to keep the numeric columns only. -> select(year, seed, reg_wins_pct, tourney_wins) |>
# Second step is to combine all numeric columns into one long column. pivot_longer(...)
# Finally, we calculate the mean, min, max of our data separately, using summarize().
# na.rm = TRUE drops the missing values, if there is any.

march_summary <- march_tidy |>
    select(year, seed, reg_wins_pct, tourney_wins) |>
    pivot_longer(cols = everything(), names_to = "variable", values_to = "value") |>
    group_by(variable) |>
    summarize(mean = mean(value, na.rm = TRUE),
              min = min(value, na.rm = TRUE),
              max = max(value, na.rm = TRUE))
march_summary
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
# is.na(.x) is to detect missing (NA) values in the current column. TRUE if missing, FALSE if not.
# sum(..) counts the TRUE results
# ~ is to apply the given function, in our case sum(is.na(.x))
# across() applies the function to every given column, year to tourney_wins in our case.
# Tried ...across(everything(), ...) -> it also works.

march_missing <- march_tidy |>
    summarize(across(year:tourney_wins, ~ sum(is.na(.x))))
march_missing
```

- Use your `dplyr` knowledge to create **one** table that calculates mean of one numeric variable for each of the levels of another categorical variable.

```r
# Our categorical variable is bid. There are two levels: auto, at-large.
# I thought about using school. But it has so many levels, so I choose bid.
# tourney_wins for the numeric variable
# group_by -> split our data into groups, summarize(mean(...)..) -> to calculate the mean

# group_by(bid) -> one for each bid type (auto and at-large)
# mean(tourney_wins) -> the mean of tournament wins for each group

march_bid_summary <- march_tidy |>
    group_by(bid) |>
    summarize(mean_tourney_wins = mean(tourney_wins))
march_bid_summary
```

```r
# For numeric variable, I am choosing tourney_wins.
# geom_histogram() will be a good choice, since it shows how often each value appears.

# ggplot(aes(x = tourney_wins)) is to put tourney_wins in x-axis
# geom_histogram(binwidth = 1, color = "black") is to draw bars that counts team for each win
# added black color to our bars, so greater visibility.
# lab(x = ..., y = ..., title = ...) is to make sure labeling is clear, reaable in English

tourney_wins_plot <- march_tidy |>
    ggplot(aes(x = tourney_wins)) + 
    geom_histogram(binwidth = 1, color = "black") +
    labs(x = "Number of Tournament Games Won", 
         y = "Number of Teams",
         title = "Distribution of Tournament Wins (1994-2018)")
tourney_wins_plot
```

```r
# calculate the proportion of teams with each bid type
# each group for bid type -> group_by(bid)
# summarize((number_of_rows = n()) -> number of teams in each group
# calculate the proportion of the total: proportion = number_of_rows / sum (number_of_rows)

# Now, plot the data
# geom_bar(stat = "identity") is to use the y-values we give.

bid_proportions <- march_tidy |>
    group_by(bid) |>
    summarize(number_of_teams = n()) |>
    mutate(proportion = number_of_teams / sum(number_of_teams))

bid_plot <- bid_proportions |>
    ggplot(aes(x = bid, y = proportion)) +
    geom_bar(stat = "identity") +
    labs(x = "Bid Type",
         y = "Proportion of Teams",
         title = "Proportion of Tournament Teams by Bid Type (1994-2018)")
bid_plot
```

Visualization 1: This plot shows the mean(average) of tournament wins for each seed. It shows a strong negative relationship. 1 seeds average is almost 4 wins, while seeds 13 to 16 range, their average is close to 0. Also, the relatiionship is curved (non linear). It drops steeply from 1 to 5 seeds and stays flat near 0 after seed 12. This suggests seed is a strong predictor of tournament wins.

```r
# Similar to Task 4.3, group_by -> summarize()
# calculate the mean of tournament wins for each seed
# Then, we plot how average tournament wins change with each seed
# label seed 1 = strongest, 16 = weakest
seed_summary <- march_tidy |>
    group_by(seed) |>
    summarize(mean_tourney_wins = mean(tourney_wins))

seed_plot <- seed_summary |>
    ggplot(aes(x = seed, y = mean_tourney_wins)) +
    geom_point(size = 3) +
    geom_line() +
    labs(x = "Seed (1 = strongest, 16 = weakest)",
         y = "Mean Number of Tournament Wins",
         title = "Mean Tournament Wins by Seed (1994-2018)")
seed_plot
```

Visualization 2: This plot shows the relationship between regular-season with percentage and the number of tournament games won. The relationship is positive but weaker than seed in Visualization 1. Teams below about 70% rarely win more than 2 games, while teams winning 5-6 games almost all have win percentages above 85%. However, the points are widely spread and many high win-percentage teams still won 0 games. This suggests win percentage is useful but a weaker predictor.

```r
# Regular season win vs tournament wins
# geom_point, since reg_wins_pct is continuous
# alpha = 0.3 -> visualization chapter. Points overlap so much -> color becomes darker
win_pct_plot <- march_tidy |>
    ggplot(aes(x = reg_wins_pct, y = tourney_wins)) +
    geom_point(alpha = 0.3) +
    labs(x = "Regular-Season Win Percentage (%)",
         y = "Number of Tournament Games Won",
         title = "Regular-Season Win Percentage vs Tournament Wins (1994-2018)")
win_pct_plot
```

I will use k-nearest neighbors (k-NN) regression because the response variable, tourney_wins is numerical. Visualization 1 showed a non-linear relationship between seed and tournament wins, and k-NN regression is appropriate because it predicts by averaging the outcomes of the most similar teams. It means it can follow a curved relationship unlike linear regression requiring a straight line.

One limitation is that k-NN regression is difficult to interpret, unlike linear regression, it doesn't produce an equation, so we can't describe how much each preditor affects the predicted number of wins. Another limitation is that most teams win 0 or 1 games (800 teams had 0 wins, while only 25 teams had 6 wins), so a strong team's nearest neighbors will likely include teams that won fewer games, causing the model to underpredict high-win teams.

To process the data, first of all, I will split it into a 75% training set and a 25% testing set before any modeling. So, the test set isn't touched and we can estimate how well the model performs on new data. Next, I will standardize both predictors using a recipe with step_scale() and step_center(). So, neither seed nor win percentage dominates the distance calculation. Then, I will choose K using 5-fold cross-validation on the training set to select the K with the lowest error, so there is no separate validation set needed. Finally, I will train the model with the best K again, on the full training set. Doing so, I can predict the test set once.

I will assess performance using RMSPE, which measures the size of prediction errors in the same units as tourney_wins (games won). It makes the model's errors easy to interpret. It is also appropriate because this is a regression problem, where accruacy can't be used since it works for classification only. Lastly, a lower RMSPE means better prediections.
